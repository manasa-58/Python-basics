1. What is hypothesis testing in statistics

   -> Hypothesis testing in statistics is a method used to evaluate claims about a population by analyzing sample data. It involves formulating a null hypothesis (a statement of no effect or no difference) and an alternative hypothesis (a statement that contradicts the null hypothesis). Researchers then collect data, calculate a test statistic, and determine whether the evidence supports rejecting the null hypothesis in favor of the alternative.

2. What is the null hypothesis, and how does it differ from the alternative hypothesis

   -> The null hypothesis is a statement of "no effect" or "no difference," while the alternative hypothesis is a statement that contradicts the null hypothesis, suggesting there is a real effect or difference. Essentially, the null hypothesis is what researchers try to disprove, while the alternative is what they aim to prove.

3. What is the significance level in hypothesis testing, and why is it important

   -> In hypothesis testing, the significance level (denoted by α, alpha) is the probability of rejecting the null hypothesis when it is actually true. It represents the researcher's willingness to accept a Type I error, which is falsely rejecting a true null hypothesis. It's a crucial threshold for determining statistical significance, helping researchers decide whether observed results are likely due to chance or a real effect.

   Controlling Type I errors:

   The significance level helps control the risk of making a Type I error. A lower significance level (e.g., 0.01) reduces the chance of a false positive but may increase the risk of failing to detect a true effect (Type II error).

   Determining statistical significance:

   It provides a threshold for determining whether observed results are statistically significant. If the p-value is below the significance level, the results are considered statistically significant, suggesting that the observed effect is unlikely due to random chance.

   Guiding decision-making:

   In research and decision-making, understanding the significance level helps researchers interpret their findings and make informed decisions.

   Balancing risk:

   Choosing the appropriate significance level involves balancing the risks of Type I and Type II errors.

4. What does a P-value represent in hypothesis testing

   -> In hypothesis testing, the P-value represents the probability of observing a test statistic as extreme as, or more extreme than, the one calculated from your sample data, assuming the null hypothesis is true. Essentially, it quantifies the likelihood that your results occurred purely by chance, under the assumption that there's no real effect or difference. A smaller P-value indicates stronger evidence against the null hypothesis.

5. How do you interpret the P-value in hypothesis testing

   -> In hypothesis testing, the p-value represents the probability of obtaining results as extreme as, or more extreme than, the observed results, assuming the null hypothesis is true. A smaller p-value indicates stronger evidence against the null hypothesis. Conventionally, if the p-value is less than or equal to a predetermined significance level (often 0.05), the null hypothesis is rejected.

6. What are Type 1 and Type 2 errors in hypothesis testing

   -> In hypothesis testing, a Type I error occurs when a true null hypothesis is incorrectly rejected, leading to a false positive. A Type II error happens when a false null hypothesis is not rejected, resulting in a false negative.

7. What is the difference between a one-tailed and a two-tailed test in hypothesis testing

   -> In hypothesis testing, a one-tailed test examines if a parameter is significantly greater or less than a specific value, while a two-tailed test examines if a parameter is significantly different from a specific value, without specifying the direction of the difference.

8. What is the Z-test, and when is it used in hypothesis testing

   -> A z-test is a statistical hypothesis test used to determine if there's a significant difference between a sample mean and a population mean, or between the means of two independent samples, when the population variance is known or the sample size is large. It's typically used when dealing with data that follows a normal distribution.

9. How do you calculate the Z-score, and what does it represent in hypothesis testing

   -> A Z-score measures how many standard deviations a data point is from the mean of a dataset. In hypothesis testing, it's used to determine the probability of obtaining a result as extreme as, or more extreme than, the one observed, assuming the null hypothesis is true. It helps determine if a sample mean is significantly different from a population mean.

   Calculating the Z-score:

   The Z-score is calculated using the following formula:

   Z = (x - μ) / σ

   Where:

   x: is the data point (or sample mean)

   μ: is the population mean

   σ: is the population standard deviation

   If you are working with a sample and don't know the population standard deviation, you can use the sample standard deviation (s) instead:

   Z = (x - x̄) / s

   Where:

   x̄: is the sample mean

   s: is the sample standard deviation

   Z-score in Hypothesis Testing:

   In hypothesis testing, a Z-test is used when you know the population standard deviation or have a large sample size. The Z-score helps determine if the observed difference between a sample mean and a hypothesized population mean is statistically significant.

10. What is the T-distribution, and when should it be used instead of the normal distribution

   -> The t-distribution, also known as Student's t-distribution, is a probability distribution used in statistics, particularly when dealing with small sample sizes or when the population standard deviation is unknown. It is preferred over the normal distribution in these situations because it accounts for the increased uncertainty associated with estimating population parameters from limited data.

11. What is the difference between a Z-test and a T-test

   -> The main difference between a Z-test and a T-test lies in the assumptions about the population standard deviation and the sample size. Z-tests are used when the population standard deviation is known and the sample size is large (typically > 30), while T-tests are used when the population standard deviation is unknown and/or the sample size is small (typically < 30).

12. What is the T-test, and how is it used in hypothesis testing

   -> A t-test is a statistical hypothesis test used to determine if there's a significant difference between the means of two groups. It's a common tool in hypothesis testing, particularly when dealing with smaller sample sizes or when the population standard deviation is unknown. The t-test helps assess whether observed differences between groups are likely due to a real effect or simply random chance.

   How t-tests are used in hypothesis testing:

   1. Formulate hypotheses:
   A null hypothesis (H0) is established, often stating that there's no significant difference between the means of the groups being compared. An alternative hypothesis (Ha) is also defined, suggesting a difference exists.
   2. Choose the appropriate t-test:
   There are different types of t-tests, including independent samples t-test (for comparing unrelated groups), paired samples t-test (for comparing related or paired data), and one-sample t-test (for comparing a sample mean to a known value).
   3. Calculate the t-statistic:
   A t-statistic is calculated using the sample data and the chosen formula for the specific t-test.
   4. Determine the p-value:
   The calculated t-statistic is used to find a p-value. The p-value represents the probability of observing the obtained results (or more extreme results) if the null hypothesis were true.
   5. Make a decision:
   If the p-value is below a predetermined significance level (alpha, usually 0.05), the null hypothesis is rejected, suggesting a significant difference between the groups. If the p-value is above the significance level, the null hypothesis is not rejected.

13. What is the relationship between Z-test and T-test in hypothesis testing

   -> Both Z-tests and T-tests are used in hypothesis testing to compare means, but they differ in their assumptions and when they are most appropriate. Z-tests are used when the population standard deviation is known and the sample size is large (typically 30 or more). T-tests are used when the population standard deviation is unknown and/or the sample size is small (typically less than 30).

14. What is a confidence interval, and how is it used to interpret statistical results

   -> A confidence interval (CI) is a range of values that is likely to contain the true value of a population parameter, such as the mean or proportion. It provides a measure of uncertainty around a sample statistic and is used to interpret statistical results by indicating the precision of an estimate and the potential range of the true population value.

15. What is the margin of error, and how does it affect the confidence interval

   -> The margin of error is a statistical measure of how much the results of a survey or study might differ from the true population value. It's a range of values, typically expressed as a plus or minus percentage, that indicates the potential variability in your results. A smaller margin of error means your results are more precise and likely closer to the true population value.

   How Margin of Error Affects Confidence Intervals:

   Confidence intervals provide a range of values within which the true population parameter is likely to fall, given a certain level of confidence.

   The margin of error is a crucial component of the confidence interval calculation.

   It essentially determines the width of the confidence interval.

   A larger margin of error results in a wider confidence interval, while a smaller margin of error leads to a narrower confidence interval.

   For example, a 95% confidence interval with a margin of error of +/- 5% means you are 95% confident that the true population value lies within 5 percentage points above or below your sample result.

   A higher confidence level (e.g., 99%) generally requires a larger margin of error to achieve the same level of certainty, resulting in a wider interval.

   Conversely, a lower confidence level (e.g., 90%) can be achieved with a smaller margin of error, leading to a narrower interval.

   Therefore, the margin of error directly influences the precision and certainty associated with the confidence interval.

16. How is Bayes' Theorem used in statistics, and what is its significance

   -> Bayes' theorem is a fundamental concept in probability and statistics that provides a way to update the probability of an event based on new evidence. It allows statisticians to combine prior knowledge with observed data to refine their understanding of probabilities, making it a crucial tool for inference and decision-making.

   Bayes' Theorem in Statistics

   Bayes' theorem is expressed mathematically as:
   P(A|B) = [P(B|A) * P(A)] / P(B)

   Significance of Bayes' Theorem
   1. Updating Probabilities:
   Bayes' theorem is significant because it allows us to update our beliefs or probabilities as new evidence becomes available. We start with a prior probability, and as we observe new data, we can calculate a posterior probability that reflects the updated likelihood of an event.
   2. Inversion of Conditional Probabilities:
   It allows us to calculate the probability of a cause (event A) given its effect (event B), even if it's easier to directly observe the probability of the effect given the cause (B|A).
   3. Bayesian Inference:
   Bayes' theorem is foundational to Bayesian inference, a statistical approach that uses Bayes' theorem to update probabilities of hypotheses or models based on observed data.
   4. Applications in Various Fields:

      Bayes' theorem has diverse applications, including:

      Medical Diagnosis: Determining the probability of a disease given symptoms or test results.

      Spam Filtering: Classifying emails as spam based on the presence of certain words.

      Machine Learning: Used in algorithms like Naive Bayes for classification and spam filtering.

      Financial Modeling: Assessing investment risks and returns.

      Risk Assessment: Evaluating the probability of various risks in different scenarios.

      Anomaly Detection: Identifying unusual patterns or outliers in datasets.

      Personalization: Recommending products or content based on user preferences.

17. What is the Chi-square distribution, and when is it used

   -> The chi-square distribution is a family of continuous probability distributions, used primarily in hypothesis testing, particularly for categorical data. It's used to determine if there's a significant difference between observed and expected frequencies, or to assess the independence of categorical variables.

18. What is the Chi-square goodness of fit test, and how is it applied

   -> The chi-square goodness-of-fit test is a statistical test used to determine if a sample distribution matches a known or expected distribution. It assesses how well observed frequencies of a categorical variable align with expected frequencies based on a theoretical distribution. Essentially, it checks if the differences between what's observed and what's expected are due to chance or if there's a significant discrepancy.

   Here's a breakdown of how it's applied:

   1. Define the Hypotheses:

     Null Hypothesis (H0): The observed data follows the expected distribution.

     Alternative Hypothesis (Ha): The observed data does not follow the expected distribution.

    2. Determine Expected Frequencies:

     For each category in your observed data, calculate the expected frequency based on the hypothesized distribution.

     If no specific distribution is hypothesized, equal probabilities (e.g., a uniform distribution) are often assumed.

   3. Calculate the Chi-Square Statistic:

     The chi-square statistic is calculated using the following formula: χ² = Σ [(Observed - Expected)² / Expected].

     Sum the results for each category.

    4. Determine Degrees of Freedom:

     Degrees of freedom (df) are calculated as (number of categories - 1).
    5. Determine the p-value:

     Using the calculated chi-square statistic and degrees of freedom, find the p-value from a chi-square distribution table or using statistical software.

     The p-value represents the probability of observing a chi-square statistic as extreme as, or more extreme than, the one calculated, if the null hypothesis is true.
    6. Make a Decision:

     Compare the p-value to a pre-determined significance level (alpha, typically 0.05).

     If the p-value is less than or equal to alpha, reject the null hypothesis and conclude that the observed distribution is significantly different from the expected distribution.

     If the p-value is greater than alpha, fail to reject the null hypothesis and conclude that there is not enough evidence to say the observed distribution is significantly different from the expected distribution.

     In simpler terms, the chi-square goodness-of-fit test helps you determine if the differences you see in your data are just random variation or if they point to a real difference between your sample and a specific distribution.

19. What is the F-distribution, and when is it used in hypothesis testing

   -> The F-distribution is a probability distribution that arises frequently in hypothesis testing, particularly when comparing variances or means of multiple groups. It is used to determine if there is a statistically significant difference between the variances of two or more populations or to test the overall significance of a regression model.

20. What is an ANOVA test, and what are its assumptions

   -> ANOVA (Analysis of Variance) is a statistical test used to compare the means of three or more groups. It determines if there are any statistically significant differences between the group means, indicating that the observed differences are likely not due to random chance. The validity of ANOVA results relies on several key assumptions.

   Assumptions of ANOVA:

   1. Normality:

     The data within each group being compared should be approximately normally distributed. This means the data should follow a bell-shaped curve with most values clustered around the mean.

  2. Homogeneity of Variance (Homoscedasticity):

     The variance (spread or dispersion) of the data should be roughly equal across all the groups being compared.

  3. Independence of Observations:

     The observations within each group and between different groups should be independent of each other. This means that the value of one data point should not influence the value of another.

     In essence, ANOVA is a powerful tool for comparing means, but it's crucial to ensure that these assumptions are met to ensure the validity of the results. Violations of these assumptions can lead to inaccurate conclusions about group differences.

21. What are the different types of ANOVA tests

   -> The primary types of ANOVA tests are one-way ANOVA and two-way ANOVA. One-way ANOVA is used to compare means of three or more groups based on a single independent variable. Two-way ANOVA, on the other hand, examines the effects of two or more independent variables on a dependent variable, and can also assess interactions between those independent variables.

22. What is the F-test, and how does it relate to hypothesis testing?

   -> The F-test is a statistical test used in hypothesis testing to compare the variances of two or more populations or samples. It is particularly useful for determining if there is a significant difference between the variances of these groups, often as a preliminary step before further analysis. The F-test relies on the F-distribution and the F-statistic, which is a ratio of two variances.

```python
#3. Write a Python function to calculate the probability distribution function (PDF) of a Bernoulli distribution

import math

def bernoulli_pdf(x, p):

    if x == 0:
        return 1 - p
    elif x == 1:
        return p
    else:
        return 0

# Example usage
probability_of_success = 0.7
outcome = 1
pdf_value = bernoulli_pdf(outcome, probability_of_success)
print(f"Probability of outcome {outcome} is: {pdf_value}")

outcome = 0
pdf_value = bernoulli_pdf(outcome, probability_of_success)
print(f"Probability of outcome {outcome} is: {pdf_value}")
```

```python
#6. Write a Python program to calculate and plot the cumulative distribution function (CDF) of a discrete uniform distribution

import numpy as np
import matplotlib.pyplot as plt

def discrete_uniform_cdf(x, a, b):

    if x < a:
        return 0.0
    elif x >= b:
        return 1.0
    else:
        # For discrete uniform distribution, the CDF is (floor(x) - a + 1) / (b - a + 1)
        # for a <= x < b
        return (np.floor(x) - a + 1) / (b - a + 1)

# Define the parameters of the discrete uniform distribution
a = 1  # Lower bound
b = 6  # Upper bound (e.g., a standard die roll)

# Generate a range of x values for plotting
x_values = np.linspace(a - 1, b + 1, 500) # Extend range for visualization

# Calculate the CDF for each x value
cdf_values = [discrete_uniform_cdf(val, a, b) for val in x_values]

# Plot the CDF
plt.figure(figsize=(10, 6))
plt.plot(x_values, cdf_values, drawstyle='steps-post', label='Discrete Uniform CDF')
plt.title(f'Cumulative Distribution Function of Discrete Uniform Distribution (a={a}, b={b})')
plt.xlabel('X')
plt.ylabel('F(X)')
plt.grid(True)
plt.yticks(np.arange(0, 1.1, 0.1)) # Set y-axis ticks for better visualization
plt.xticks(np.arange(a - 1, b + 2, 1)) # Set x-axis ticks
plt.legend()
plt.show()
```

```python
#9. Write a Python function to calculate Z-scores from a dataset and plot them

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import zscore # For a more robust Z-score calculation

def calculate_and_plot_zscores(data):

    # Convert data to a NumPy array for efficient calculations
    data_array = np.array(data)

    # Calculate Z-scores
    # Z = (x - mean) / standard_deviation
    # Using scipy.stats.zscore is recommended for robustness
    z_scores = zscore(data_array)

    # Plotting the Z-scores
    plt.figure(figsize=(8, 6))
    plt.hist(z_scores, bins=20, edgecolor='black', alpha=0.7)
    plt.title('Distribution of Z-Scores')
    plt.xlabel('Z-Score')
    plt.ylabel('Frequency')
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.axvline(0, color='red', linestyle='--', label='Mean (Z=0)') # Mark the mean Z-score
    plt.legend()
    plt.show()

    print("Calculated Z-Scores:")
    print(z_scores)

# Example usage:
if __name__ == "__main__":
    sample_data = [10, 12, 12, 13, 15, 16, 18, 20, 22, 25, 30, 50]
    calculate_and_plot_zscores(sample_data)

    another_data = np.random.normal(loc=100, scale=15, size=500) # Normally distributed data
    calculate_and_plot_zscores(another_data)
```

```python
#14. Write a Python program to calculate the Z-score for a given data point and compare it to a standard normal distribution

import numpy as np
from scipy.stats import norm

def calculate_z_score(data_point, data_set):

    mean = np.mean(data_set)
    std_dev = np.std(data_set)
    z_score = (data_point - mean) / std_dev
    return z_score

def compare_to_standard_normal(z_score):

    p_value = norm.cdf(z_score)
    return p_value

# Example Usage
if __name__ == "__main__":
    # Sample data
    data = [10, 12, 15, 12, 18, 20, 16, 14, 11, 19]
    target_data_point = 18

    # Calculate Z-score
    z = calculate_z_score(target_data_point, data)
    print(f"The Z-score for {target_data_point} is: {z:.2f}")

    # Compare to standard normal distribution
    probability = compare_to_standard_normal(z)
    print(f"The cumulative probability (p-value) for this Z-score is: {probability:.4f}")

    # Interpretation
    print("\nInterpretation:")
    if probability < 0.05 or probability > 0.95:
        print("This data point is relatively far from the mean, potentially an outlier.")
    else:
        print("This data point is relatively close to the mean.")
```

```python
#20. Simulate a random variable using a continuous uniform distribution and calculate its expected value

import random
import numpy as np

def simulate_uniform(a, b, n):
    """
    Simulates n random variables from a uniform distribution [a, b]
    and calculates the expected value.
    """
    # Generate random numbers
    random_numbers = [a + (b - a) * random.random() for _ in range(n)]

    # Calculate the expected value
    expected_value = (a + b) / 2

    # Calculate the sample mean
    sample_mean = np.mean(random_numbers)

    return random_numbers, expected_value, sample_mean

# Example usage:
a = 2
b = 10
n = 10000

random_numbers, expected_value, sample_mean = simulate_uniform(a, b, n)

print(f"Expected value: {expected_value}")
print(f"Sample mean: {sample_mean}")
print(f"Difference: {abs(expected_value - sample_mean)}")

# Verification:  The difference between the expected value and the sample mean should be small, especially for larger n.
```

```python
#Practical Part - 2
#1. Write a Python program to perform a Z-test for comparing a sample mean to a known population mean and interpret the results

import numpy as np
import scipy.stats as st
from statsmodels.stats.weightstats import ztest

def perform_z_test(sample_data, population_mean, population_std_dev, alpha=0.05, alternative='two-sided'):

    # Calculate the sample mean
    sample_mean = np.mean(sample_data)
    n = len(sample_data)

    # Calculate the z-statistic
    z_statistic = (sample_mean - population_mean) / (population_std_dev / np.sqrt(n))

    # Calculate the p-value
    if alternative == 'two-sided':
        p_value = 2 * (1 - st.norm.cdf(np.abs(z_statistic)))
    elif alternative == 'larger':
        p_value = 1 - st.norm.cdf(z_statistic)
    elif alternative == 'smaller':
        p_value = st.norm.cdf(z_statistic)
    else:
        raise ValueError("Invalid alternative hypothesis. Choose 'two-sided', 'larger', or 'smaller'.")

    # Interpret the results
    if p_value < alpha:
        result = "Reject the null hypothesis. There is a significant difference."
    else:
        result = "Fail to reject the null hypothesis. There is no significant difference."

    return z_statistic, p_value, result

# Example usage
sample_data = [25, 28, 26, 30, 27, 29, 28]
population_mean = 20
population_std_dev = 5
alpha = 0.05

z_statistic, p_value, result = perform_z_test(sample_data, population_mean, population_std_dev, alpha=alpha)

print(f"Z-Statistic: {z_statistic:.4f}")
print(f"P-value: {p_value:.4f}")
print(result)

#Example with alternative 'larger'
z_statistic, p_value, result = perform_z_test(sample_data, population_mean, population_std_dev, alpha=alpha, alternative='larger')

print(f"\nZ-Statistic (alternative='larger'): {z_statistic:.4f}")
print(f"P-value (alternative='larger'): {p_value:.4f}")
print(result)
```

```python
#3. Implement a one-sample Z-test using Python to compare the sample mean with the population mean

import numpy as np
from scipy import stats

def one_sample_z_test(sample_data, population_mean, population_std_dev, alpha=0.05, alternative='two-sided'):

    sample_mean = np.mean(sample_data)
    sample_size = len(sample_data)

    # Calculate the Z-statistic
    z_statistic = (sample_mean - population_mean) / (population_std_dev / np.sqrt(sample_size))

    # Calculate the p-value based on the alternative hypothesis
    if alternative == 'two-sided':
        p_value = 2 * (1 - stats.norm.cdf(abs(z_statistic)))
    elif alternative == 'less':
        p_value = stats.norm.cdf(z_statistic)
    elif alternative == 'greater':
        p_value = 1 - stats.norm.cdf(z_statistic)
    else:
        raise ValueError("Invalid 'alternative' argument. Must be 'two-sided', 'less', or 'greater'.")

    # Make the decision
    if p_value < alpha:
        decision = "Reject the null hypothesis. The sample mean is significantly different from the population mean."
    else:
        decision = "Fail to reject the null hypothesis. There is no significant difference between the sample mean and the population mean."

    return {
        "Z-statistic": z_statistic,
        "P-value": p_value,
        "Decision": decision
    }

# Example Usage:
# Suppose we have a sample of student test scores and want to compare it to a known population mean.
sample_scores = [85, 90, 78, 92, 88, 75, 80, 95, 87, 83]
known_population_mean = 80
known_population_std_dev = 10

results = one_sample_z_test(sample_scores, known_population_mean, known_population_std_dev)
print(results)

# Example with a different alternative hypothesis
results_greater = one_sample_z_test(sample_scores, known_population_mean, known_population_std_dev, alternative='greater')
print(results_greater)
```

```python
#4. Perform a two-tailed Z-test using Python and visualize the decision region on a plot

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# --- 1. Define Hypotheses and Parameters ---
# Example: Testing if a sample mean differs significantly from a hypothesized population mean.
population_mean_hypothesized = 100
population_std_dev = 15 # Assumed known population standard deviation
sample_size = 50
sample_mean = 105
alpha = 0.05 # Significance level for a two-tailed test

# --- 2. Calculate Z-statistic ---
standard_error = population_std_dev / np.sqrt(sample_size)
z_statistic = (sample_mean - population_mean_hypothesized) / standard_error

# --- 3. Determine Critical Z-values ---
# For a two-tailed test, alpha is split into two tails (alpha/2)
critical_z_positive = norm.ppf(1 - alpha / 2)
critical_z_negative = norm.ppf(alpha / 2)

# --- 4. Calculate P-value ---
p_value = 2 * (1 - norm.cdf(abs(z_statistic)))

# --- 5. Make a Decision ---
print(f"Calculated Z-statistic: {z_statistic:.2f}")
print(f"P-value: {p_value:.4f}")
print(f"Critical Z-values: ±{critical_z_positive:.2f}")

if abs(z_statistic) > critical_z_positive:
    print("Decision: Reject the Null Hypothesis (H0).")
else:
    print("Decision: Fail to Reject the Null Hypothesis (H0).")

# --- 6. Visualize the Decision Region ---
x = np.linspace(-4, 4, 500)
y = norm.pdf(x)

plt.figure(figsize=(10, 6))
plt.plot(x, y, color='blue', label='Standard Normal Distribution')

# Shade the rejection regions
plt.fill_between(x, 0, y, where=(x >= critical_z_positive) | (x <= critical_z_negative),
                 color='red', alpha=0.3, label='Rejection Region')

# Plot the calculated Z-statistic
plt.axvline(z_statistic, color='green', linestyle='--', label=f'Calculated Z-statistic: {z_statistic:.2f}')

# Add critical Z-value lines
plt.axvline(critical_z_positive, color='purple', linestyle=':', label=f'Critical Z-value: {critical_z_positive:.2f}')
plt.axvline(critical_z_negative, color='purple', linestyle=':')

plt.title('Two-tailed Z-test: Decision Region Visualization')
plt.xlabel('Z-score')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#5. Create a Python function that calculates and visualizes Type 1 and Type 2 errors during hypothesis testing

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

def visualize_type_errors(mean_null, std_null, mean_alt, alpha, beta=None, sample_size=100):

    # Generate data for the null and alternative distributions
    x_null = np.linspace(mean_null - 4 * std_null, mean_null + 4 * std_null, sample_size)
    y_null = norm.pdf(x_null, mean_null, std_null)
    x_alt = np.linspace(mean_alt - 4 * std_null, mean_alt + 4 * std_null, sample_size)
    y_alt = norm.pdf(x_alt, mean_alt, std_null)


    # Calculate critical value for the null distribution
    critical_value = norm.ppf(1 - alpha, loc=mean_null, scale=std_null)

    # Calculate Type I error (alpha)
    type_1_error = norm.sf(critical_value, loc=mean_null, scale=std_null)

    # Calculate Type II error (beta)
    if beta is None:
        beta = norm.cdf(critical_value, loc=mean_alt, scale=std_null)


    # Plotting
    plt.figure(figsize=(12, 6))

    # Null distribution
    plt.plot(x_null, y_null, label=f'Null Hypothesis (μ={mean_null}, σ={std_null})')
    plt.fill_between(x_null[x_null >= critical_value], 0, y_null[x_null >= critical_value], color='red', alpha=0.3, label=f'Type I Error (α = {type_1_error:.2f})')
    plt.axvline(x=critical_value, color='red', linestyle='--', label=f'Critical Value (x={critical_value:.2f})')

    # Alternative distribution
    plt.plot(x_alt, y_alt, label=f'Alternative Hypothesis (μ={mean_alt}, σ={std_null})')
    plt.fill_between(x_alt[x_alt <= critical_value], 0, y_alt[x_alt <= critical_value], color='orange', alpha=0.3, label=f'Type II Error (β = {beta:.2f})')
    plt.xlabel("Values")
    plt.ylabel("Probability Density")
    plt.title("Type I and Type II Errors in Hypothesis Testing")
    plt.legend()
    plt.grid(True)
    plt.show()


# Example usage:
mean_null = 0
std_null = 1
mean_alt = 1.5
alpha = 0.05  # Significance level
visualize_type_errors(mean_null, std_null, mean_alt, alpha)
```

```python
#6. Write a Python program to perform an independent T-test and interpret the results

from scipy import stats
import numpy as np

def perform_independent_t_test(group1_data, group2_data, alpha=0.05):

    # Perform the independent t-test
    # equal_var=True assumes equal variances, set to False for Welch's t-test if unequal variances are suspected
    t_statistic, p_value = stats.ttest_ind(group1_data, group2_data, equal_var=True)

    interpretation = ""
    if p_value < alpha:
        interpretation = (
            f"Reject the null hypothesis (p < {alpha}). "
            "There is a statistically significant difference between the means of the two groups."
        )
    else:
        interpretation = (
            f"Fail to reject the null hypothesis (p >= {alpha}). "
            "There is no statistically significant difference between the means of the two groups."
        )

    return t_statistic, p_value, interpretation

# Example Usage:
# Let's say we have scores of two different teaching methods
method_a_scores = [85, 88, 90, 82, 79, 91, 87, 84, 89, 86]
method_b_scores = [78, 80, 75, 83, 77, 81, 79, 76, 82, 79]

t_stat, p_val, interpretation_result = perform_independent_t_test(method_a_scores, method_b_scores)

print(f"Independent Samples T-Test Results:")
print(f"T-statistic: {t_stat:.3f}")
print(f"P-value: {p_val:.3f}")
print(f"Interpretation: {interpretation_result}")
```

```python
#9. Write a Python function to calculate the confidence interval for a sample mean and explain its significance

import numpy as np
import scipy.stats as st

def confidence_interval(data, confidence=0.95):
    """
    Calculates the confidence interval for a sample mean.

    Args:
        data: A list or NumPy array of numerical data.
        confidence: The desired confidence level (default is 0.95 for 95%).

    Returns:
        A tuple containing the lower and upper bounds of the confidence interval.
    """
    a = 1.0 * np.array(data)
    n = len(a)
    if n == 0:
        return (None, None)
    m, se = np.mean(a), st.sem(a)
    h = se * st.t.ppf((1 + confidence) / 2., n-1)
    return m-h, m+h

#10. Write a Python program to calculate the margin of error for a given confidence level using sample data

import numpy as np
from scipy import stats

def calculate_margin_of_error(data, confidence_level=0.95):

    sample_mean = np.mean(data)
    sample_std = np.std(data, ddof=1)  # Use ddof=1 for sample standard deviation
    sample_size = len(data)

    # Calculate the critical t-value
    # alpha is the significance level, which is 1 - confidence_level
    # We need to find the t-value for (1 + confidence_level) / 2 because stats.t.ppf
    # gives the value for a single tail
    alpha = 1 - confidence_level
    t_critical = stats.t.ppf(1 - alpha / 2, df=sample_size - 1)

    # Calculate the standard error of the mean
    standard_error = sample_std / np.sqrt(sample_size)

    # Calculate the margin of error
    margin_of_error = t_critical * standard_error

    return margin_of_error

# Example usage:
sample_data = [25, 28, 22, 30, 27, 26, 29, 24, 31, 23]
confidence_level_95 = 0.95
confidence_level_99 = 0.99

moe_95 = calculate_margin_of_error(sample_data, confidence_level_95)
moe_99 = calculate_margin_of_error(sample_data, confidence_level_99)

print(f"Sample Data: {sample_data}")
print(f"Margin of Error (95% Confidence): {moe_95:.2f}")
print(f"Margin of Error (99% Confidence): {moe_99:.2f}")
```

```python
#11. Implement a Bayesian inference method using Bayes' Theorem in Python and explain the process

def bayes_theorem(prior_A, likelihood_B_given_A, likelihood_B_given_not_A):
    """
    Calculates the posterior probability using Bayes' Theorem.

    Args:
      prior_A: The prior probability of event A.
      likelihood_B_given_A: The likelihood of event B given A.
      likelihood_B_given_not_A: The likelihood of event B given not A.

    Returns:
      The posterior probability of event A given event B.
    """
    # Calculate the probability of not A
    not_A = 1 - prior_A

    # Calculate the marginal probability of B (evidence)
    # P(B) = P(B|A) * P(A) + P(B|~A) * P(~A)
    evidence_B = (likelihood_B_given_A * prior_A) + (likelihood_B_given_not_A * not_A)

    # Calculate the posterior probability of A given B
    # P(A|B) = (P(B|A) * P(A)) / P(B)
    posterior_A_given_B = (likelihood_B_given_A * prior_A) / evidence_B

    return posterior_A_given_B

# Example usage
# Let's say we want to determine the probability that a person has a disease (A)
# given they tested positive for it (B).
# Prior probability of having the disease: 1%
prior_disease = 0.01
# Likelihood of a positive test given the person has the disease: 99%
likelihood_positive_given_disease = 0.99
# Likelihood of a positive test given the person does not have the disease: 5% (false positive)
likelihood_positive_given_no_disease = 0.05

posterior_disease_given_positive = bayes_theorem(prior_disease,
                                                 likelihood_positive_given_disease,
                                                 likelihood_positive_given_no_disease)

print(f"The posterior probability of having the disease given a positive test is: {posterior_disease_given_positive:.4f}")

# Bayesian updating example.  Suppose we observe a second positive test.
likelihood_positive_given_disease_2nd_test = 0.99
likelihood_positive_given_no_disease_2nd_test = 0.05

#Treat the previous posterior as the new prior for the second test
new_prior_disease = posterior_disease_given_positive
new_posterior_disease_given_positive = bayes_theorem(new_prior_disease,
                                                 likelihood_positive_given_disease_2nd_test,
                                                 likelihood_positive_given_no_disease_2nd_test)
print(f"The posterior probability of having the disease given two positive tests is: {new_posterior_disease_given_positive:.4f}")
```

```python
#12. Perform a Chi-square test for independence between two categorical variables in Python

import pandas as pd
from scipy.stats import chi2_contingency

# Sample data (replace with your actual data)
data = {'Variable1': ['A', 'A', 'B', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
        'Variable2': ['X', 'Y', 'X', 'Y', 'X', 'X', 'Y', 'Y', 'X', 'Y']}
df = pd.DataFrame(data)

# Create a contingency table
contingency_table = pd.crosstab(df['Variable1'], df['Variable2'])
print("Contingency Table:\n", contingency_table)

# Perform the Chi-square test
chi2_stat, p_value, dof, expected_frequencies = chi2_contingency(contingency_table)

print(f"\nChi-square Statistic: {chi2_stat}")
print(f"P-value: {p_value}")
print(f"Degrees of Freedom: {dof}")
print("Expected Frequencies:\n", expected_frequencies)

# Interpret the p-value
alpha = 0.05
if p_value <= alpha:
    print("\nResult: Reject the null hypothesis. There is a significant association between the variables.")
else:
    print("\nResult: Fail to reject the null hypothesis. The variables are independent.")
```

```python
#13. Write a Python program to calculate the expected frequencies for a Chi-square test based on observed data

import numpy as np

def calculate_expected_frequencies(observed_data):

    row_totals = np.sum(observed_data, axis=1)  # Sum across rows to get row totals
    col_totals = np.sum(observed_data, axis=0)  # Sum across columns to get column totals
    grand_total = np.sum(observed_data)         # Sum of all observations

    # Calculate expected frequencies using the formula: (row_total * col_total) / grand_total
    expected_frequencies = np.outer(row_totals, col_totals) / grand_total
    return expected_frequencies

# Example Usage:
# Observed data for a 2x2 contingency table
# e.g., Gender vs. Political Party Preference
# Rows: Male, Female
# Columns: Republican, Democrat
observed_data_example = np.array([[50, 30],
                                  [40, 60]])

expected_freq = calculate_expected_frequencies(observed_data_example)
print("Observed Frequencies:\n", observed_data_example)
print("\nExpected Frequencies:\n", expected_freq)
```

```python
#14. Perform a goodness-of-fit test using Python to compare the observed data to an expected distribution.

import numpy as np
from scipy import stats

# Observed frequencies
observed_data = np.array([25, 30, 20, 15, 10])

# Expected frequencies (e.g., assuming a uniform distribution)
# The sum of expected frequencies should equal the sum of observed frequencies
total_observations = np.sum(observed_data)
num_categories = len(observed_data)
expected_data = np.array([total_observations / num_categories] * num_categories)

# Perform the Chi-Square Goodness-of-Fit Test
chi_square_statistic, p_value = stats.chisquare(f_obs=observed_data, f_exp=expected_data)

print(f"Observed Data: {observed_data}")
print(f"Expected Data: {expected_data}")
print(f"Chi-Square Statistic: {chi_square_statistic}")
print(f"P-value: {p_value}")

# Interpretation based on a significance level of 0.05
alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis: The observed distribution is significantly different from the expected distribution.")
else:
    print("Fail to reject the null hypothesis: The observed distribution does not significantly differ from the expected distribution.")
```
