#1. What is Anomaly Detection? Explain its types (point, contextual, and collective anomalies) with examples.

   -> Anomaly detection is the process of identifying rare data points, events, or patterns that deviate significantly from the norm and raise suspicions, often signaling a problem like fraud or a system failure. There are three main types of anomalies: Point Anomalies are single data points that are unusual compared to the rest of the dataset. Contextual Anomalies are data points that become anomalous due to the specific context in which they occur, even if they would be normal in another context. Collective Anomalies are a collection of data points that, when considered together, form an unusual pattern, even though individual points might not be anomalous.

Types of Anomalies

1. Point Anomalies

   What it is: A single data point that is significantly different from the majority of the data.

   Example: In a dataset of human heights, a single person who is 8 feet tall would be a point anomaly, as it is far outside the typical range for human height.

2. Contextual Anomalies

   What it is: A data point that is not inherently unusual but becomes anomalous when considered within its specific context.

   Example: A temperature reading of 30°C (86°F) is normal in summer but would be considered a contextual anomaly in winter, especially in a cold climate.

3. Collective Anomalies

   What it is: A group of data points that, when viewed together, reveal an unusual pattern, even if the individual points are not anomalous.

   Example: A series of frequent, small credit card transactions that, on their own, seem legitimate. However, when viewed as a group and happening in a rapid succession, they form a collective anomaly that could indicate fraudulent activity aimed at evading detection.

#2. Compare Isolation Forest, DBSCAN, and Local Outlier Factor in terms of their approach and suitable use cases.

   -> Isolation Forest builds random trees to isolate outliers quickly in high-dimensional, large datasets by giving anomalies shorter paths. DBSCAN is a density-based algorithm that groups dense data points into clusters and flags outliers as noise in low-density regions. Local Outlier Factor (LOF) is a density-based anomaly detection method that identifies outliers by comparing a point's local density to its neighbors, making it suitable for data with varying local densities.

Isolation Forest

Approach:
Builds an ensemble of decision trees to isolate outliers. Anomalies are points that require fewer random splits to be isolated.

Suitable Use Cases:
High-dimensional data where other methods might struggle with the curse of dimensionality.
Large datasets due to its speed and scalability.
When global outliers are more of a concern.

DBSCAN

Approach:
A density-based clustering algorithm that identifies clusters of high-density points and labels low-density points as noise (outliers).

Suitable Use Cases:
Identifying anomalies in noisy datasets.
Discovering clusters of arbitrary shapes and structures.
Situations where the number of clusters is unknown.

Local Outlier Factor (LOF)

Approach:
Measures the local density deviation of a data point compared to its neighbors. Outliers are points with significantly lower densities than their surrounding points.

Suitable Use Cases:
Detecting local anomalies and variations in density across the dataset.
Datasets where clusters have varying densities, a situation where global methods like Isolation Forest might not perform as well.
Situations where computing pairwise distances is feasible, as it's computationally more expensive than Isolation Forest for large datasets.   

#3. What are the key components of a Time Series? Explain each with one example.

   -> A time series has four key components: Trend, the overall long-term movement of the data; Seasonality, short-term, regular patterns that repeat within a year; Cyclicity, longer-term, irregular, and non-fixed patterns that span multiple years; and Irregularity (or Noise), unpredictable, random fluctuations not accounted for by the other components. For example, e-commerce sales might show an upward trend over years, higher seasonal sales in December, a slower recovery period (a cycle) during a business downturn, and a sudden drop due to a delivery disruption (an irregularity).

Here's a breakdown of each component:

1. Trend

   Explanation: This is the long-term upward, downward, or static direction of the time series data over an extended period.

   Example: The overall increase in global e-commerce sales over the past decade is an example of an upward trend.

2. Seasonal

   Explanation: This refers to patterns that repeat at a fixed interval within a year, such as daily, weekly, or monthly cycles.

   Example: Electricity usage typically peaks during summer months due to increased air conditioner use and declines during other seasons, showing a seasonal pattern with a one-year period.

3. Cyclical

   Explanation: These are longer-term fluctuations that are not of a fixed period or duration, often associated with business cycles or economic expansions and recessions.

   Example: An ice cream company might observe higher sales in summer and lower sales in winter, but also a less frequent pattern of overall sales decline over several years (a business cycle) followed by a period of recovery.

4. Irregular (Noise)

   Explanation: This component comprises random, unpredictable fluctuations or "noise" that cannot be explained by trend, seasonal, or cyclical patterns.

   Example: A sudden, temporary drop in website traffic due to an unexpected server outage or a localized power cut would be considered an irregular component.

#4. Define Stationary in time series. How can you test and transform a non-stationary series into a stationary one?

   -> A time series is stationary if its statistical properties, such as mean, variance, and autocorrelation, do not change over time. To test for stationarity, you can use the Augmented Dickey-Fuller (ADF) test and observe summary statistics and run sequence plots. To transform a non-stationary series, you can apply transformations like differencing (to remove trends) and logarithms (to stabilize variance).

What is a Stationary Time Series?

A stationary time series is one whose statistical properties are constant over time. This means:

Constant Mean:
The average value of the series remains consistent.

Constant Variance:
The spread or volatility of the data doesn't change.

Constant Autocorrelation:
The correlation between an observation and its past values depends only on the time lag between them, not on the specific time point.

How to Test for Stationarity

You can test for stationarity using both visual and statistical methods:

Visual Inspection:
Run Sequence Plot: A time series plot can reveal trends (upward or downward slopes) or seasonality (repeating patterns) that indicate non-stationarity.

Histograms: Look for shifts in the distribution of the data.

Statistical Tests:

Augmented Dickey-Fuller (ADF) Test: This is a common statistical test that checks for the presence of a unit root in a time series. A unit root is a characteristic of non-stationary time series. If the ADF test's p-value is less than a chosen significance level (e.g., 0.05), you reject the null hypothesis and conclude the series is stationary.

How to Transform a Non-Stationary Series

If a series is non-stationary, you can make it stationary using transformations:

Differencing: This involves subtracting the previous observation from the current one (e.g., Yt - Yt-1). Differencing helps to:
Remove trends by stabilizing the mean.
Remove seasonality by taking differences over the appropriate seasonal period (e.g., a yearly difference to remove yearly seasonality).

Logging: Applying a logarithmic transformation can help to stabilize the variance of a time series if the variance increases over time.
After applying these transformations, you should re-test the series for stationarity to confirm that the transformation was successful.

#5. Differentiate between AR, MA, ARIMA, SARIMA, and SARIMAX models in terms of structure and application.

   -> AR (Autoregressive) and MA (Moving Average) are basic components of time series models, while ARIMA (Autoregressive Integrated Moving Average) handles non-seasonal, non-stationary data by differencing the time series. SARIMA (Seasonal ARIMA) adds seasonal components to ARIMA for data with recurring patterns, and SARIMAX (Seasonal ARIMA with eXogenous variables) further enhances SARIMA by incorporating external factors.

AR (Autoregressive) Model

Structure:
Predicts future values based on a linear combination of past values of the series. Represented as AR(p), where 'p' is the order (number of past values included).

Application:
Used when the time series exhibits a strong relationship with its past values, but the data must be stationary (mean and variance are constant over time).

MA (Moving Average) Model

Structure:
Predicts future values based on past errors (the difference between actual and predicted values). Represented as MA(q), where 'q' is the order of past errors.

Application:
Suitable for stationary time series that show patterns in past prediction errors, but it doesn't account for long-term trends or seasonality.

ARIMA (Autoregressive Integrated Moving Average) Model

Structure:
Combines AR and MA components with an integration (I) component, which involves differencing the time series to achieve stationarity. An ARIMA(p, d, q) model has an AR order 'p', an integration order 'd' (number of differences), and an MA order 'q'.

Application:
Ideal for forecasting non-seasonal time series that are not stationary but can be made stationary through differencing.

SARIMA (Seasonal ARIMA) Model

Structure:
Extends ARIMA by adding seasonal components. It includes non-seasonal orders (p, d, q) and seasonal orders (P, D, Q, s), where 's' is the length of the seasonal period.

Application:
Best for time series data that exhibits both non-seasonal trends and recurring seasonal patterns, like monthly sales data or daily temperature readings.

SARIMAX (Seasonal ARIMA with eXogenous Variables) Model
Structure:
An enhancement of SARIMA that incorporates one or more external variables (exogenous regressors) that influence the time series.
Application:
When external factors, such as marketing spend affecting sales or weather affecting tourism, are believed to impact the time series, SARIMAX can provide more accurate forecasts by accounting for these influences.


   



```python
#6. Load a time series dataset (e.g., AirPassengers), plot the original series, and decompose it into trend, seasonality, and residual components

import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

# 1. Load the dataset
# The AirPassengers dataset is often available online or can be loaded from a local CSV.
# For demonstration, we'll use a common URL for the dataset.
url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/airline-passengers.csv"
df = pd.read_csv(url, parse_dates=['Month'], index_col='Month')
series = df['Passengers']

# 2. Plot the original series
plt.figure(figsize=(10, 6))
plt.plot(series, label='Original Air Passengers Series')
plt.title('Air Passengers Time Series Data')
plt.xlabel('Date')
plt.ylabel('Number of Passengers')
plt.legend()
plt.grid(True)
plt.show()

# 3. Decompose the series
# Choose 'additive' or 'multiplicative' model based on whether seasonality
# increases with the trend (multiplicative) or remains constant (additive).
# AirPassengers typically exhibits multiplicative seasonality.
decomposition = seasonal_decompose(series, model='multiplicative', period=12) # period=12 for monthly data

# Plot the decomposed components
fig = decomposition.plot()
fig.set_size_inches(10, 8)
plt.suptitle('Decomposition of Air Passengers Time Series', y=1.02) # Adjust suptitle position
plt.tight_layout(rect=[0, 0, 1, 0.98]) # Adjust layout to prevent suptitle overlap
plt.show()
```

```python
#7. Apply Isolation Forest on a numerical dataset (e.g., NYC Taxi Fare) to detect anomalies. Visualize the anomalies on a 2D scatter plot.

import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

# Simulate NYC Taxi Fare data
np.random.seed(42)
# Normal data points
fare_amount = np.random.normal(loc=15, scale=5, size=200)
trip_distance = np.random.normal(loc=3, scale=1, size=200)

# Introduce some anomalies
anomalous_fare = np.random.uniform(low=50, high=100, size=10)
anomalous_distance = np.random.uniform(low=10, high=20, size=10)

X = np.column_stack((np.concatenate((fare_amount, anomalous_fare)),
                      np.concatenate((trip_distance, anomalous_distance))))


# Instantiate Isolation Forest model
# contamination is the proportion of outliers in the data set
model = IsolationForest(contamination=0.05, random_state=42)
model.fit(X)

    # Predict anomalies
y_pred = model.predict(X)

    # Visualize the results
plt.figure(figsize=(10, 6))
plt.scatter(X[y_pred == 1, 0], X[y_pred == 1, 1], c='blue', label='Normal Points', alpha=0.6)
plt.scatter(X[y_pred == -1, 0], X[y_pred == -1, 1], c='red', label='Anomalies', s=100, edgecolors='k')
plt.title('Anomaly Detection using Isolation Forest on NYC Taxi Fare Data')
plt.xlabel('Fare Amount')
plt.ylabel('Trip Distance')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#8. Train a SARIMA model on the monthly airline passengers dataset. Forecast the next 12 months and visualize the results.

import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm

# Load the dataset (assuming 'AirPassengers.csv' is available)
data = pd.read_csv('AirPassengers.csv', index_col='Month', parse_dates=True)
data.index = pd.to_datetime(data.index, format='%Y-%m')
data.index.freq = 'MS' # Set frequency to Month Start

# Rename column for clarity
data.columns = ['Passengers']

    # Train the SARIMA model
# (p,d,q) = (1,1,1) for non-seasonal, (P,D,Q,s) = (1,1,1,12) for seasonal
model = sm.tsa.SARIMAX(data['Passengers'],
                        order=(1, 1, 1),
                        seasonal_order=(1, 1, 1, 12),
                        enforce_stationarity=False,
                        enforce_invertibility=False)
results = model.fit(disp=False)
print(results.summary())

    # Forecast the next 12 months
forecast_steps = 12
forecast = results.get_forecast(steps=forecast_steps)
forecast_ci = forecast.conf_int()

    # Plot the original data and the forecast
plt.figure(figsize=(12, 6))
plt.plot(data.index, data['Passengers'], label='Original Data')
plt.plot(forecast.predicted_mean.index, forecast.predicted_mean, label='Forecast')
plt.fill_between(forecast_ci.index,
                  forecast_ci.iloc[:, 0],
                  forecast_ci.iloc[:, 1], color='k', alpha=.15, label='95% Confidence Interval')
plt.title('Monthly Airline Passengers Forecast with SARIMA')
plt.xlabel('Date')
plt.ylabel('Number of Passengers')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#9. Apply Local Outlier Factor (LOF) on any numerical dataset to detect anomalies and visualize them using matplotlib.

import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import LocalOutlierFactor

# 1. Generate a sample numerical dataset with outliers
np.random.seed(42) # for reproducibility
X = 0.3 * np.random.randn(100, 2)
X_outliers = np.random.uniform(low=-4, high=4, size=(10, 2))
X = np.r_[X + 2, X - 2, X_outliers]

# 2. Apply Local Outlier Factor (LOF)
# n_neighbors: number of neighbors to consider for local density calculation
# contamination: proportion of outliers in the dataset (estimated)
lof = LocalOutlierFactor(n_neighbors=20, contamination=0.1)
y_pred = lof.fit_predict(X)

# The LOF score is the negative of the outlier factor,
# where lower values indicate higher likelihood of being an outlier.
# We can access the raw outlier scores using lof.negative_outlier_factor_
# For visualization, we'll use -lof.negative_outlier_factor_ to have higher values for outliers.
outlier_scores = -lof.negative_outlier_factor_

# 3. Visualize the results
plt.figure(figsize=(10, 7))

# Plot normal points in blue
plt.scatter(X[y_pred == 1, 0], X[y_pred == 1, 1], color='blue', s=50, label='Normal Points')

# Plot outliers in red, with size proportional to outlier score
# Scale the size for better visualization
s = 200 * (outlier_scores - outlier_scores.min()) / (outlier_scores.max() - outlier_scores.min()) + 50
plt.scatter(X[y_pred == -1, 0], X[y_pred == -1, 1], color='red', s=s[y_pred == -1], label='Outliers')

plt.title("Local Outlier Factor (LOF) Anomaly Detection")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True)
plt.show()
```
