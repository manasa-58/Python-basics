# Statistics Advance Part 1

1. What is a random variable in probability theory

   -> In probability, a real-valued function, defined over the sample space of a random experiment, is called a random variable. That is, the values of the random variable correspond to the outcomes of the random experiment. Random variables could be either discrete or continuous.

2. What are the types of random variables

   -> There are two main types of random variables: discrete and continuous. Discrete random variables can only take on a finite number of values or a countably infinite number of values, like integers. Continuous random variables can take on any value within a given range.

3. What is the difference between discrete and continuous distributions

   -> A discrete distribution is one in which the data can only take on certain values, for example integers. A continuous distribution is one in which data can take on any value within a specified range (which may be infinite).

4. What are probability distribution functions (PDF)

   -> A probability distribution is a mathematical function that describes the probability of different possible values of a variable. Probability distributions are often depicted using graphs or probability tables.

5. How do cumulative distribution functions (CDF) differ from probability distribution functions (PDF)

   -> The Probability Density Function (PDF) and Cumulative Distribution Function (CDF) are both functions used to describe the distribution of a random variable, but they represent different aspects. The PDF gives the probability density at a specific point, while the CDF gives the probability that a random variable is less than or equal to a specific value.

6. What is a discrete uniform distribution

   -> A discrete uniform distribution is a probability distribution where a finite number of outcomes are all equally likely. Think of rolling a fair six-sided die; each number (1 through 6) has a probability of 1/6 of being rolled, making it a discrete uniform distribution.

7. What are the key properties of a Bernoulli distribution

   -> A Bernoulli distribution models a single trial with only two possible outcomes: success (usually denoted as 1) or failure (usually denoted as 0). The key properties include: it's a discrete distribution, it's binary, the probability of success is constant across trials, and the trials are independent. The mean is 'p' (the probability of success), and the variance is p(1-p).

8. What is the binomial distribution, and how is it used in probability

   -> The binomial distribution is a probability distribution that models the probability of obtaining a specific number of successes in a fixed number of independent trials, where each trial has only two possible outcomes (success or failure) and the probability of success is constant across all trials. It's used in probability to analyze situations with binary outcomes, like coin flips, defect rates, or survey responses.

9. What is the Poisson distribution and where is it applied

   -> The Poisson distribution is a discrete probability distribution that expresses the probability of a given number of events occurring in a fixed interval of time or space if these events occur with a known constant mean rate and independently of the time since the last event. It's used when you're dealing with rare events that happen independently at a constant average rate within a specific interval.

10. What is a continuous uniform distribution

   -> A continuous uniform distribution is a probability distribution where a continuous random variable is equally likely to take any value within a specified range (a, b). This means the probability of the variable falling within any sub-interval of the same width is the same.

11. What are the characteristics of a normal distribution

   -> A normal distribution, often called a bell curve, is characterized by its symmetry, unimodal shape (single peak), and the fact that its mean, median, and mode are all equal and located at the center. The curve is asymptotic, meaning its tails approach the x-axis but never touch it. Additionally, the spread of the distribution is determined by the standard deviation, and the empirical rule helps predict the distribution of data within certain standard deviations of the mean.

12. What is the standard normal distribution, and why is it important

   -> The standard normal distribution is a specific type of normal distribution with a mean of 0 and a standard deviation of 1. It's a crucial concept in statistics because it allows for the comparison of data from different normal distributions and enables probability calculations using z-scores. By converting data into z-scores (number of standard deviations from the mean), you can use a standard normal table to find probabilities associated with specific values.

13. What is the Central Limit Theorem (CLT), and why is it critical in statistics

   -> The Central Limit Theorem (CLT) states that the distribution of sample means will approach a normal distribution, regardless of the original population distribution, as the sample size becomes sufficiently large. This is critical in statistics because it allows us to use the normal distribution and its associated properties to make inferences about populations, even when we don't know the population's true distribution.

14. How does the Central Limit Theorem relate to the normal distribution

   -> The central limit theorem (CLT) essentially states that the distribution of sample means will approximate a normal distribution, regardless of the underlying population distribution, as long as the sample size is sufficiently large. This means that even if you are sampling from a population that is not normally distributed, the averages of multiple samples will tend towards a normal distribution.

15. What is the application of Z statistics in hypothesis testing

   -> Z-statistics, calculated using a z-test, are applied in hypothesis testing to determine if there's a significant difference between a sample mean and a population mean, or between the means of two samples, when certain conditions are met. Specifically, z-tests are used when the population variance is known, or when the sample size is large (typically n ≥ 30). The z-statistic helps assess the likelihood of observing the sample data if the null hypothesis is true.

16. How do you calculate a Z-score, and what does it represent

   -> A z-score represents the number of standard deviations a data point is away from the mean of a dataset. It's calculated by subtracting the mean from the data point and dividing the result by the standard deviation. A positive z-score indicates the data point is above the mean, while a negative z-score indicates it's below the mean.

17. What are point estimates and interval estimates in statistics

   -> In statistics, point estimates and interval estimates are two methods used to estimate unknown population parameters from sample data. Point estimates provide a single value as the best guess for the parameter, while interval estimates provide a range of values that are likely to contain the parameter, along with a level of confidence.

18. What is the significance of confidence intervals in statistical analysis

   -> Confidence intervals are crucial in statistical analysis because they provide a range of plausible values for an unknown population parameter, indicating the uncertainty associated with sample estimates. They help researchers understand the reliability and precision of their findings, allowing for more informed conclusions and decisions.

19. What is the relationship between a Z-score and a confidence interval

   -> A Z-score and a confidence interval are related because the Z-score is used to calculate the margin of error, which is a key component of a confidence interval. Specifically, the Z-score corresponding to a chosen confidence level is used to determine how many standard deviations away from the sample mean the confidence interval's boundaries should be.

20. How are Z-scores used to compare different distributions

   -> Z-scores are used to compare data from different distributions by standardizing the values. They represent how many standard deviations a data point is from its distribution's mean, allowing for direct comparison even if the distributions have different means and standard deviations.

21. What are the assumptions for applying the Central Limit Theorem

   -> The central limit theorem (CLT) allows us to assume that the distribution of sample means will be approximately normal, regardless of the population's distribution, as long as the sample size is sufficiently large. This is crucial for many statistical inferences. The main assumptions are: random sampling, independence of samples, and sufficiently large sample size.

22. What is the concept of expected value in a probability distribution

   -> The expected value in a probability distribution represents the long-run average outcome of a random variable. It's a weighted average of all possible values, where each value is weighted by its corresponding probability. Essentially, it tells you what you can "expect" to happen on average if you repeat an experiment many times.

23. How does a probability distribution relate to the expected outcome of a random variable?

   -> A probability distribution describes the likelihood of different outcomes for a random variable, and the expected outcome (or expected value) is a weighted average of those outcomes, where the weights are their respective probabilities. In essence, the probability distribution tells you what could happen, and the expected value gives you a sense of the average outcome you might expect over many trials.


```python
#1. Write a Python program to generate a random variable and display its value
import random

# Generate a random integer between 1 and 10 (inclusive)
random_integer = random.randint(1, 10)

# Generate a random floating-point number between 0.0 and 1.0 (exclusive of 1.0)
random_float = random.random()

# Display the generated values
print(f"Generated random integer: {random_integer}")
print(f"Generated random float: {random_float}")

#Output: Generated random integer: 10
#Output: Generated random float: 0.1555403305031946

#2. Generate a discrete uniform distribution using Python and plot the probability mass function (PMF)
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import randint

# Define the range for the discrete uniform distribution (inclusive lower, exclusive upper)
# For example, a standard six-sided die would be low=1, high=7
low_bound = 1
high_bound = 7  # Represents outcomes 1, 2, 3, 4, 5, 6

# Create a discrete uniform random variable object
discrete_uniform_dist = randint(low=low_bound, high=high_bound)

# Get the possible outcomes (values) for the PMF
# np.arange creates an array from low_bound up to (but not including) high_bound
outcomes = np.arange(low_bound, high_bound)

# Calculate the Probability Mass Function (PMF) for each outcome
# For a discrete uniform distribution, each outcome has an equal probability
pmf_values = discrete_uniform_dist.pmf(outcomes)

# Plot the PMF
plt.figure(figsize=(8, 6))
plt.bar(outcomes, pmf_values, width=0.8, align='center', color='skyblue')
plt.xlabel('Outcome')
plt.ylabel('Probability')
plt.title(f'Probability Mass Function of a Discrete Uniform Distribution (Range: {low_bound} to {high_bound-1})')
plt.xticks(outcomes) # Ensure all outcomes are shown on the x-axis
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()


#3. Write a Python function to calculate the probability distribution function (PDF) of a Bernoulli distribution
import numpy as np

def bernoulli_pdf(x, p):

    if not 0 <= p <= 1:
        raise ValueError("Probability 'p' must be between 0 and 1.")

    if isinstance(x, (int, float)):
        if x not in (0, 1):
             raise ValueError("x must be 0 or 1 for Bernoulli distribution")
        return p**x * (1-p)**(1-x)
    else:
        x = np.array(x)
        if not np.all((x == 0) | (x == 1)):
             raise ValueError("x must contain only 0 or 1 for Bernoulli distribution")
        return p**x * (1-p)**(1-x)

#4. Write a Python script to simulate a binomial distribution with n=10 and p=0.5, then plot its histogram
import numpy as np
import matplotlib.pyplot as plt

# Define the parameters for the binomial distribution
n = 10  # Number of trials
p = 0.5  # Probability of success

# Generate random samples from the binomial distribution
# 'size' determines how many times the experiment (10 trials with p=0.5) is repeated.
# A larger 'size' will result in a histogram that more closely resembles the theoretical distribution.
num_samples = 10000
data = np.random.binomial(n=n, p=p, size=num_samples)

# Create the histogram
plt.hist(data, bins=np.arange(-0.5, n + 1.5, 1), density=True, edgecolor='black', alpha=0.7)

# Add labels and title
plt.title(f'Simulated Binomial Distribution (n={n}, p={p})')
plt.xlabel('Number of Successes')
plt.ylabel('Probability Density')
plt.xticks(range(n + 1)) # Ensure x-axis ticks are at integer values representing number of successes
plt.grid(True, linestyle='--', alpha=0.6)

# Display the plot
plt.show()

#5. Create a Poisson distribution and visualize it using Python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Define the lambda (mean rate of events) for the Poisson distribution
lam = 3

# Generate random samples from a Poisson distribution
# 'size' determines the number of samples to generate
data = np.random.poisson(lam=lam, size=1000)

# Visualize the distribution using a histogram
plt.figure(figsize=(8, 6))
sns.histplot(data, bins=np.arange(-0.5, max(data) + 1.5, 1), kde=False, color='skyblue', edgecolor='black')
plt.title(f'Poisson Distribution (λ={lam})')
plt.xlabel('Number of Events')
plt.ylabel('Frequency')
plt.xticks(np.arange(0, max(data) + 1, 1)) # Ensure integer ticks on x-axis
plt.grid(True)
plt.show()

#6. Write a Python program to calculate and plot the cumulative distribution function (CDF) of a discrete uniform distribution
import numpy as np
import matplotlib.pyplot as plt

def discrete_uniform_cdf(x, a, b):
    """
    Calculates the CDF of a discrete uniform distribution.

    Args:
        x (int or float): The value at which to evaluate the CDF.
        a (int): The lower bound of the discrete uniform distribution (inclusive).
        b (int): The upper bound of the discrete uniform distribution (inclusive).

    Returns:
        float: The CDF value at x.
    """
    if not (isinstance(a, int) and isinstance(b, int) and a <= b):
        raise ValueError("a and b must be integers with a <= b")

    if x < a:
        return 0.0
    elif x >= b:
        return 1.0
    else:
        # For a discrete uniform distribution, the CDF is (floor(x) - a + 1) / (b - a + 1)
        # for x within the range [a, b).
        return (np.floor(x) - a + 1) / (b - a + 1)

# Define the parameters of the discrete uniform distribution
a = 1  # Lower bound
b = 10 # Upper bound

# Create a range of x values to plot the CDF
x_values = np.arange(a - 2, b + 3, 0.1) # Extend range slightly for visualization

# Calculate the CDF for each x value
cdf_values = [discrete_uniform_cdf(val, a, b) for val in x_values]

# Plot the CDF
plt.figure(figsize=(10, 6))
plt.step(x_values, cdf_values, where='post', label=f'Discrete Uniform CDF ({a} to {b})')
plt.title('Cumulative Distribution Function of a Discrete Uniform Distribution')
plt.xlabel('X')
plt.ylabel('P(X <= x)')
plt.grid(True)
plt.legend()
plt.show()

#7. Generate a continuous uniform distribution using NumPy and visualize it
import numpy as np
import matplotlib.pyplot as plt
# Generate 10000 samples from a uniform distribution between 0 and 10
samples = np.random.uniform(low=0, high=10, size=10000)

plt.figure(figsize=(8, 6))
plt.hist(samples, bins=50, density=True, edgecolor='black', alpha=0.7)
plt.title('Continuous Uniform Distribution')
plt.xlabel('Value')
plt.ylabel('Density')
plt.grid(True)
plt.show()

#8. Simulate data from a normal distribution and plot its histogram
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Parameters for the normal distribution
mean = 0
std_dev = 1
num_samples = 1000

# Generate random samples
data = np.random.normal(mean, std_dev, num_samples)

# Create the histogram
plt.hist(data, bins=30, density=True, alpha=0.6, color='skyblue', label='Histogram')

# Overlay the normal distribution curve
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = norm.pdf(x, mean, std_dev)
plt.plot(x, p, 'k', linewidth=2, label='Normal Distribution')

# Add labels and title
plt.xlabel('Value')
plt.ylabel('Frequency/Probability')
plt.title('Normal Distribution Histogram')
plt.legend()
plt.grid(True)
plt.show()

#9. Write a Python function to calculate Z-scores from a dataset and plot them
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import zscore

def calculate_and_plot_zscores(data, title="Z-Scores of Dataset"):
    """
    Calculates Z-scores for a given dataset and plots them.

    Args:
        data (list or np.array): The input dataset.
        title (str): The title for the plot.
    """
    # Convert data to a NumPy array if it's not already
    data = np.array(data)

    # Calculate Z-scores
    z_scores = zscore(data)

    # Create the plot
    plt.figure(figsize=(10, 6))
    plt.stem(range(len(z_scores)), z_scores, basefmt=" ", linefmt="b-", markerfmt="bo")
    plt.axhline(0, color='gray', linestyle='--', linewidth=0.8) # Add a horizontal line at Z=0
    plt.xlabel("Data Point Index")
    plt.ylabel("Z-Score")
    plt.title(title)
    plt.grid(True)
    plt.show()

    return z_scores

# Example Usage:
if __name__ == "__main__":
    # Sample dataset
    dataset = [10, 12, 11, 15, 8, 20, 9, 100, 13, 14]

    # Calculate and plot Z-scores
    calculated_z_scores = calculate_and_plot_zscores(dataset, "Z-Scores of Sample Data")

    print("\nCalculated Z-scores:")
    print(calculated_z_scores)

#10. Implement the Central Limit Theorem (CLT) using Python for a non-normal distribution.
import numpy as np
import matplotlib.pyplot as plt

# Generate data from an exponential distribution (non-normal)
population = np.random.exponential(scale=2.0, size=100000)

sample_size = 50
num_samples = 1000

sample_means = []
for _ in range(num_samples):
    sample = np.random.choice(population, size=sample_size)
    sample_means.append(np.mean(sample))

#11. Simulate multiple samples from a normal distribution and verify the Central Limit Theorem
import numpy as np
import matplotlib.pyplot as plt

# Set parameters
num_samples = 10000
sample_size = 30
original_distribution_mean = 5
original_distribution_std = 2

# Create a skewed distribution (e.g., exponential)
# You can change this to other distributions (uniform, etc.)
original_distribution = np.random.exponential(scale=original_distribution_mean, size=num_samples)

# Store sample means
sample_means = []

# Simulate multiple samples and calculate their means
for _ in range(num_samples):
    # Generate a sample from the original distribution
    sample = np.random.choice(original_distribution, size=sample_size, replace=True)
    # Calculate the mean of the sample
    sample_mean = np.mean(sample)
    # Store the sample mean
    sample_means.append(sample_mean)

# Plot the original distribution and the distribution of sample means
plt.figure(figsize=(12, 5))

# Original distribution
plt.subplot(1, 2, 1)
plt.hist(original_distribution, bins=50, density=True, alpha=0.6, color='blue', label='Original Distribution')
plt.title('Original Distribution (e.g., Exponential)')
plt.xlabel('Value')
plt.ylabel('Density')
plt.legend()

# Distribution of sample means
plt.subplot(1, 2, 2)
plt.hist(sample_means, bins=50, density=True, alpha=0.6, color='red', label='Sample Means')
plt.title('Distribution of Sample Means (CLT)')
plt.xlabel('Mean Value')
plt.ylabel('Density')

# Overlay a normal distribution for comparison
x = np.linspace(min(sample_means), max(sample_means), 100)
plt.plot(x, 1/(np.std(sample_means) * np.sqrt(2 * np.pi)) * np.exp( - (x - np.mean(sample_means))**2 / (2 * np.std(sample_means)**2) ),
         linewidth=2, color='green', label='Normal Distribution')
plt.legend()

plt.tight_layout()
plt.show()

#12. Write a Python function to calculate and plot the standard normal distribution (mean = 0, std = 1)
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

def plot_standard_normal_distribution(x_range=(-4, 4), num_points=1000):
    """
    Calculates and plots the standard normal distribution (mean=0, std=1).

    Args:
        x_range (tuple): A tuple (min_x, max_x) defining the range for the x-axis.
        num_points (int): The number of points to generate for the plot.
    """
    # Generate x values within the specified range
    x = np.linspace(x_range[0], x_range[1], num_points)

    # Calculate the PDF for the standard normal distribution
    # norm.pdf(x, loc=mean, scale=std_dev)
    # For standard normal distribution, loc=0 and scale=1
    pdf_values = norm.pdf(x, loc=0, scale=1)

    # Plotting
    plt.figure(figsize=(8, 6))
    plt.plot(x, pdf_values, color='blue', label='Standard Normal Distribution (μ=0, σ=1)')
    plt.title('Standard Normal Distribution')
    plt.xlabel('X')
    plt.ylabel('Probability Density')
    plt.grid(True)
    plt.legend()
    plt.show()

# Example usage:
# plot_standard_normal_distribution()
# plot_standard_normal_distribution(x_range=(-5, 5), num_points=500)

#13. Generate random variables and calculate their corresponding probabilities using the binomial distribution
import math

def binomial_probability(n, k, p):
  """
  Calculates the binomial probability.

  Args:
    n: The number of trials.
    k: The number of successes.
    p: The probability of success in a single trial.

  Returns:
    The binomial probability P(X=k).
  """
  combinations = math.comb(n, k) # or math.factorial(n) // (math.factorial(k) * math.factorial(n - k))
  probability = combinations * (p**k) * ((1-p)**(n-k))
  return probability

# Example usage:
n_trials = 5
num_successes = 3
probability_of_success = 0.5

probability = binomial_probability(n_trials, num_successes, probability_of_success)
print(f"The probability of {num_successes} successes in {n_trials} trials is: {probability}")

#Output: The probability of 3 successes in 5 trials is: 0.3125

#14. Write a Python program to calculate the Z-score for a given data point and compare it to a standard normal distribution
import numpy as np
from scipy.stats import norm

def calculate_z_score(data_point, data):
    """
    Calculates the Z-score for a given data point within a dataset.

    Args:
        data_point (float): The individual data point for which to calculate the Z-score.
        data (list or numpy.ndarray): The dataset containing the data point.

    Returns:
        float: The calculated Z-score.
    """
    mean = np.mean(data)
    std_dev = np.std(data)
    z_score = (data_point - mean) / std_dev
    return z_score

def compare_to_standard_normal(z_score):
    """
    Compares a calculated Z-score to a standard normal distribution.

    Args:
        z_score (float): The Z-score to compare.

    Returns:
        tuple: A tuple containing:
            - float: The cumulative probability (P-value) corresponding to the Z-score.
            - str: An interpretation of the Z-score relative to the standard normal distribution.
    """
    p_value = norm.cdf(z_score)

    if z_score > 0:
        interpretation = f"The data point is {abs(z_score):.2f} standard deviations above the mean."
    elif z_score < 0:
        interpretation = f"The data point is {abs(z_score):.2f} standard deviations below the mean."
    else:
        interpretation = "The data point is exactly at the mean."

    return p_value, interpretation

# Example Usage
if __name__ == "__main__":
    # Sample data
    data = [6, 7, 7, 12, 13, 13, 15, 16, 19, 22]

    # Data point to analyze
    my_data_point = 19

    # Calculate Z-score
    z = calculate_z_score(my_data_point, data)
    print(f"The Z-score for {my_data_point} is: {z:.2f}")

    # Compare to standard normal distribution
    p, interpretation_text = compare_to_standard_normal(z)
    print(f"Cumulative probability (P-value): {p:.4f}")
    print(interpretation_text)

    # Another example
    another_data_point = 7
    z_another = calculate_z_score(another_data_point, data)
    print(f"\nThe Z-score for {another_data_point} is: {z_another:.2f}")
    p_another, interpretation_another = compare_to_standard_normal(z_another)
    print(f"Cumulative probability (P-value): {p_another:.4f}")
    print(interpretation_another)

    """ Output: The Z-score for 19 is: 1.20
Cumulative probability (P-value): 0.8840
The data point is 1.20 standard deviations above the mean.

The Z-score for 7 is: -1.20
Cumulative probability (P-value): 0.1160
The data point is 1.20 standard deviations below the mean.
"""

#15. Implement hypothesis testing using Z-statistics for a sample dataset

"""

Let's say you want to test if the average height of women in your city is 5'4" (162.56 cm), and you have a sample of 100 women with a sample mean height of 5'5"
(165.1 cm) and a known population standard deviation of 2 inches (5.08 cm).

Hypotheses:
H0: μ = 5'4"
H1: μ ≠ 5'4"
Significance Level: α = 0.05

Z-statistic: z = z = (x̄ - μ) / (σ / √n)
Z-statistic: z = (5'5" - 5'4") / (2" / √100) = 5 / (2 / 10) = 25

"""

#16. Create a confidence interval for a dataset using Python and interpret the result

import numpy as np
from scipy import stats

# Sample data
data = np.array([22, 25, 20, 28, 23, 26, 21, 24, 27, 22])

# Calculate sample statistics
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1) # ddof=1 for sample standard deviation
n = len(data)
degrees_freedom = n - 1

# Set confidence level
confidence_level = 0.95

# Calculate the confidence interval
confidence_interval = stats.t.interval(
    confidence_level,
    degrees_freedom,
    loc=sample_mean,
    scale=sample_std / np.sqrt(n)
)

print(f"Sample Mean: {sample_mean:.2f}")
print(f"{int(confidence_level*100)}% Confidence Interval: {confidence_interval}")

#Output: Sample Mean: 23.80
#Output: 95% Confidence Interval: (np.float64(21.898352235346195), np.float64(25.701647764653806))

#17. Generate data from a normal distribution, then calculate and interpret the confidence interval for its mean

import numpy as np
import scipy.stats as st

# 1. Generate data from a normal distribution
np.random.seed(42)  # for reproducibility
data = np.random.normal(loc=10, scale=2, size=30)  # Sample of 30 from N(10, 2)
# print(data)

# 2. Calculate sample statistics
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)  # Use ddof=1 for sample standard deviation
n = len(data)
# print(f"Sample Mean: {sample_mean:.2f}")
# print(f"Sample Standard Deviation: {sample_std:.2f}")
# print(f"Sample size: {n}")


# 3. Choose a confidence level and calculate the margin of error
confidence_level = 0.95
alpha = 1 - confidence_level
# For a large sample (n > 30), use the z-distribution. For smaller samples, use the t-distribution
# Here, we assume a large sample
z_critical = st.norm.ppf(1 - alpha / 2)
# print(f"Z-critical value: {z_critical:.2f}")

# Calculate the standard error
standard_error = sample_std / np.sqrt(n)
# print(f"Standard Error: {standard_error:.2f}")

# Calculate the margin of error
margin_of_error = z_critical * standard_error
# print(f"Margin of Error: {margin_of_error:.2f}")


# 4. Calculate the confidence interval
confidence_interval = (sample_mean - margin_of_error, sample_mean + margin_of_error)

print(f"Sample Mean: {sample_mean:.2f}")
print(f"Confidence Interval (95%): {confidence_interval}")


# 5. Interpretation
print(f"\nWe are {confidence_level*100}% confident that the true population mean lies between {confidence_interval[0]:.2f} and {confidence_interval[1]:.2f}.")

# Example with a smaller sample size to demonstrate t-distribution
data_small = np.random.normal(loc=10, scale=2, size=10)
sample_mean_small = np.mean(data_small)
sample_std_small = np.std(data_small, ddof=1)
n_small = len(data_small)

# Calculate t-critical value
t_critical_small = st.t.ppf(1 - alpha / 2, n_small - 1)
# print(f"T-critical value (small sample): {t_critical_small:.2f}")

# Calculate standard error
standard_error_small = sample_std_small / np.sqrt(n_small)
# print(f"Standard Error (small sample): {standard_error_small:.2f}")

# Calculate margin of error
margin_of_error_small = t_critical_small * standard_error_small
# print(f"Margin of Error (small sample): {margin_of_error_small:.2f}")

# Calculate confidence interval
confidence_interval_small = (sample_mean_small - margin_of_error_small, sample_mean_small + margin_of_error_small)

print(f"\nSample Mean (small sample): {sample_mean_small:.2f}")
print(f"Confidence Interval (small sample, 95%): {confidence_interval_small}")
print(f"We are {confidence_level*100}% confident that the true population mean lies between {confidence_interval_small[0]:.2f} and {confidence_interval_small[1]:.2f}.")

""" Output: Sample Mean: 9.62
Confidence Interval (95%): (np.float64(8.979591716611449), np.float64(10.267820699984405))

We are 95.0% confident that the true population mean lies between 8.98 and 10.27.

Sample Mean (small sample): 9.38
Confidence Interval (small sample, 95%): (np.float64(7.741283656799602), np.float64(11.018289678725814))
We are 95.0% confident that the true population mean lies between 7.74 and 11.02.

"""
#18. Write a Python script to calculate and visualize the probability density function (PDF) of a normal distribution

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Define parameters for the normal distribution
mean = 0  # Mean (mu) of the distribution
std_dev = 1  # Standard deviation (sigma) of the distribution

# Generate a range of x values for the plot
# We typically go a few standard deviations away from the mean to capture most of the distribution
x = np.linspace(mean - 4 * std_dev, mean + 4 * std_dev, 500)

# Calculate the PDF for each x value using the norm.pdf function
pdf_values = norm.pdf(x, loc=mean, scale=std_dev)

# Create the plot
plt.figure(figsize=(8, 6))
plt.plot(x, pdf_values, color='blue', label=f'Normal Distribution (μ={mean}, σ={std_dev})')
plt.title('Probability Density Function of a Normal Distribution')
plt.xlabel('X')
plt.ylabel('Probability Density')
plt.grid(True)
plt.legend()
plt.show()

#19. Use Python to calculate and interpret the cumulative distribution function (CDF) of a Poisson distribution

import numpy as np
from scipy.stats import poisson
import matplotlib.pyplot as plt

def poisson_cdf(lam, k):
    """
    Calculates the cumulative distribution function (CDF) of a Poisson distribution.

    Args:
    lam (float): The lambda parameter of the Poisson distribution (average rate).
    k (int): The number of events (non-negative integer).

    Returns:
    float: The CDF value, representing the probability of observing k or fewer events.
    """
    return poisson.cdf(k, lam)

def plot_poisson_cdf(lam, max_k=20):
    """
    Plots the CDF of a Poisson distribution.

    Args:
    lam (float): The lambda parameter of the Poisson distribution.
    max_k (int): The maximum number of events to plot.
    """
    k_values = np.arange(0, max_k + 1)
    cdf_values = [poisson_cdf(lam, k) for k in k_values]

    plt.step(k_values, cdf_values, where='post')
    plt.xlabel("Number of Events (k)")
    plt.ylabel("Cumulative Probability")
    plt.title(f"Poisson CDF (λ = {lam})")
    plt.grid(True)
    plt.show()

# Example usage
lambda_val = 3  # Example: Average of 3 events
k_value = 5  # Calculate CDF for 5 or fewer events

# Calculate CDF for a specific k
cdf_result = poisson_cdf(lambda_val, k_value)
print(f"CDF at k = {k_value} for λ = {lambda_val}: {cdf_result}")

# Plot the CDF
plot_poisson_cdf(lambda_val)


#20. Simulate a random variable using a continuous uniform distribution and calculate its expected value

import numpy as np

# Define the interval
a = 2
b = 8

# Simulate a random value
random_value = np.random.uniform(low=a, high=b)
print(f"Simulated random value: {random_value}")

# Calculate the expected value
expected_value = (a + b) / 2
print(f"Expected value: {expected_value}")

#Output: Simulated random value: 7.817507766587351
#Output: Expected value: 5.0


#21. Write a Python program to compare the standard deviations of two datasets and visualize the difference

import numpy as np
import matplotlib.pyplot as plt

def compare_and_visualize_std_dev(data1, data2, label1="Dataset 1", label2="Dataset 2"):
    """
    Compares the standard deviations of two datasets and visualizes the difference.

    Args:
        data1 (list or np.array): The first dataset.
        data2 (list or np.array): The second dataset.
        label1 (str): Label for the first dataset in the plot.
        label2 (str): Label for the second dataset in the plot.
    """

    # Calculate standard deviations
    std_dev1 = np.std(data1)
    std_dev2 = np.std(data2)

    print(f"Standard Deviation of {label1}: {std_dev1:.4f}")
    print(f"Standard Deviation of {label2}: {std_dev2:.4f}")

    # Create a bar plot to visualize the difference
    labels = [label1, label2]
    std_devs = [std_dev1, std_dev2]

    plt.figure(figsize=(8, 6))
    plt.bar(labels, std_devs, color=['skyblue', 'lightcoral'])
    plt.ylabel('Standard Deviation')
    plt.title('Comparison of Standard Deviations')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()

# Example Usage:
if __name__ == "__main__":
    # Sample datasets
    dataset_a = [10, 12, 11, 13, 10, 14, 12, 11]
    dataset_b = [5, 15, 8, 18, 6, 17, 9, 16]

    # Compare and visualize
    compare_and_visualize_std_dev(dataset_a, dataset_b, "Temperatures City A", "Temperatures City B")

    # Another example with different data
    dataset_c = np.random.normal(loc=50, scale=5, size=100)
    dataset_d = np.random.normal(loc=50, scale=15, size=100)
    compare_and_visualize_std_dev(dataset_c, dataset_d, "Scores Group X", "Scores Group Y")


#22. Calculate the range and interquartile range (IQR) of a dataset generated from a normal distribution

"""

Let's say your normally distributed dataset is: 10, 12, 15, 18, 20, 22, 25, 28, 30.

Range: The range is the difference between the maximum and minimum values in the dataset.
Range: 30 - 10 = 20

Q2 (median): 20
Q1 (median of lower half): (12 + 15) / 2 = 13.5
Q3 (median of upper half): (25 + 28) / 2 = 26.5
IQR: Q3 - Q1
IQR: 26.5 - 13.5 = 13

"""

#23. Implement Z-score normalization on a dataset and visualize its transformation

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import zscore

# Sample data (replace with your actual data)
data = np.array([10, 12, 15, 18, 20, 22, 25, 28, 30, 32, 35])

# Calculate mean and standard deviation
mean = np.mean(data)
std_dev = np.std(data)

# Z-score normalization
z_scores = zscore(data)

# Alternatively, manual calculation
# z_scores_manual = (data - mean) / std_dev

# Visualization
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.hist(data, bins=5, edgecolor='black')
plt.title('Original Data')
plt.xlabel('Value')
plt.ylabel('Frequency')

plt.subplot(1, 2, 2)
plt.hist(z_scores, bins=5, edgecolor='black')
plt.title('Z-Score Normalized Data')
plt.xlabel('Z-Score')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()

# Check if normalization was successful (optional)
print(f"Mean of original data: {mean}")
print(f"Standard deviation of original data: {std_dev}")
print(f"Mean of z-score normalized data: {np.mean(z_scores):.2f}")
print(f"Standard deviation of z-score normalized data: {np.std(z_scores):.2f}")

#24. Write a Python function to calculate the skewness and kurtosis of a dataset generated from a normal distribution.

import numpy as np
from scipy.stats import skew, kurtosis

def calculate_skewness_kurtosis(data):
  """
  Calculates the skewness and kurtosis of a dataset.

  Args:
      data: A NumPy array or list representing the dataset.

  Returns:
      A tuple containing the skewness and kurtosis values, respectively.
  """
  return skew(data), kurtosis(data)


def generate_normal_distribution_data(mean=0, std_dev=1, size=1000):
    """
    Generates a dataset from a normal distribution.

    Args:
        mean: The mean of the normal distribution.
        std_dev: The standard deviation of the normal distribution.
        size: The number of data points to generate.

    Returns:
        A NumPy array containing the generated data.
    """
    return np.random.normal(mean, std_dev, size)

if __name__ == '__main__':
    # Example usage
    # Generate a sample dataset from a normal distribution
    data = generate_normal_distribution_data()

    # Calculate skewness and kurtosis
    skewness, kurtosis_val = calculate_skewness_kurtosis(data)

    # Print the results
    print(f"Skewness: {skewness}")
    print(f"Kurtosis: {kurtosis_val}")

#Output: Skewness: -0.040174358023533335
#Output: Kurtosis: -0.057746238206544565



```
